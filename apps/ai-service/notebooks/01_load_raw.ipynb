# 01 Load Raw Synthea Data

This notebook loads the raw Synthea CSV files without applying domain transformations. It is the raw ingestion layer for inspection, debugging, and re-import support.

```python
from pathlib import Path
import pandas as pd

RAW_DATA_DIR = Path('data/raw')

CSV_FILES = {
    'patients': RAW_DATA_DIR / 'patients.csv',
    'observations': RAW_DATA_DIR / 'observations.csv',
    'conditions': RAW_DATA_DIR / 'conditions.csv',
    'medications': RAW_DATA_DIR / 'medications.csv',
    'encounters': RAW_DATA_DIR / 'encounters.csv',
}

CSV_FILES
```

```python
tables = {}

for name, path in CSV_FILES.items():
    if path.exists():
        tables[name] = pd.read_csv(path)
        print(f'Loaded {name}: {tables[name].shape}')
    else:
        print(f'Missing file: {path}')
```

```python
tables.keys()
```

```python
for name, df in tables.items():
    print('\n' + '=' * 80)
    print(name.upper())
    print('=' * 80)
    print(df.head())
    print('\nColumns:')
    print(df.columns.tolist())
    print('\nDtypes:')
    print(df.dtypes)
```

```python
for name, df in tables.items():
    print('\n' + '-' * 80)
    print(f'{name} null counts')
    print('-' * 80)
    print(df.isna().sum().sort_values(ascending=False).head(20))
```
