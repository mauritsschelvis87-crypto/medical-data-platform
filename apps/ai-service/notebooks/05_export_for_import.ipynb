# 05 Validate and Export for Backend Import

This notebook validates the processed datasets and prepares an export package for the backend import layer.

```python
import shutil
from pathlib import Path
import pandas as pd

PROCESSED_DATA_DIR = Path('data/processed')
FEATURES_DATA_DIR = Path('data/features')
EXPORT_DATA_DIR = Path('data/exports')

EXPORT_DATA_DIR.mkdir(parents=True, exist_ok=True)

PATIENT_FILE = PROCESSED_DATA_DIR / 'patient.csv'
PATIENT_ADDRESS_FILE = PROCESSED_DATA_DIR / 'patient_address.csv'
VITAL_SIGNS_FILE = PROCESSED_DATA_DIR / 'vital_signs.csv'
PATIENT_FEATURES_FILE = FEATURES_DATA_DIR / 'patient_features.csv'
```

```python
patient_df = pd.read_csv(PATIENT_FILE)
patient_address_df = pd.read_csv(PATIENT_ADDRESS_FILE)
vital_signs_df = pd.read_csv(VITAL_SIGNS_FILE)
patient_features_df = pd.read_csv(PATIENT_FEATURES_FILE)

print('Patient shape:', patient_df.shape)
print('Patient address shape:', patient_address_df.shape)
print('Vital signs shape:', vital_signs_df.shape)
print('Patient features shape:', patient_features_df.shape)
```

```python
print('Duplicate patient ids:', patient_df['id'].duplicated().sum())
print('Duplicate patient numbers:', patient_df['patient_number'].duplicated().sum())
print('Addresses with missing patient FK:', (~patient_address_df['patient_id'].isin(patient_df['id'])).sum())
print('Vital signs with missing patient FK:', (~vital_signs_df['patient_id'].isin(patient_df['id'])).sum())
print('Feature rows with missing patient FK:', (~patient_features_df['patient_id'].isin(patient_df['id'])).sum())
```

```python
required_patient_columns = ['id', 'patient_number', 'source_patient_id', 'first_name', 'last_name', 'birth_date', 'gender']
required_address_columns = ['id', 'patient_id', 'address_line', 'city', 'state', 'county', 'zip_code']
required_vital_columns = ['id', 'patient_id', 'vital_type', 'value', 'unit', 'measured_at']

print('Missing patient columns:', [c for c in required_patient_columns if c not in patient_df.columns])
print('Missing patient_address columns:', [c for c in required_address_columns if c not in patient_address_df.columns])
print('Missing vital_signs columns:', [c for c in required_vital_columns if c not in vital_signs_df.columns])
```

```python
export_files = [PATIENT_FILE, PATIENT_ADDRESS_FILE, VITAL_SIGNS_FILE, PATIENT_FEATURES_FILE]

for file_path in export_files:
    target_path = EXPORT_DATA_DIR / file_path.name
    shutil.copy(file_path, target_path)
    print(f'Copied {file_path} -> {target_path}')
```
